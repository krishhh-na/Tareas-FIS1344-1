```python
import numpy as np                              #importamos bibliotecas
import matplotlib.pyplot as plt

def f(t,y):                                     #definimos la función
  return 10*y - 10/t**2 - 2/t**3

def rkf(f, a, b, y0, tol, hmin=1e-8, hmax=0.25):     #definimos RKF, establecemos un valor minimo y maximo
    t = a                                            #valores iniciales
    w = y0
    h = hmax

    ts = [t]                                   #listas para guardar resultados
    sols = [w]

    while t < b - 1e-12:
        h = min(h, b-t)
        k1 = h*f(t, w)                                                              #evaluaciones de la funcion
        k2 = h*f(t + h/4, w + k1/4)
        k3 = h*f(t + 3*h/8, w + 3*k1/32 + 9*k2/32)
        k4 = h*f(t + 12*h/13, w + 1932*k1/2197 - 7200*k2/2197 + 7296*k3/2197)
        k5 = h*f(t + h, w + 439*k1/216 - 8*k2 + 3680*k3/513 - 845*k4/4104)
        k6 = h*f(t + h/2, w - 8*k1/27 + 2*k2 - 3544*k3/2565 + 1859*k4/4104 - 11*k5/40)

        R = abs(k1/360 - 128*k3/4275 - 2197*k4/75240 + k5/50 + 2*k6/55) / h          #estima del error normalizado por h

        if R <= tol:                                                                #si el error es menor que la tolerancia, se acepta el paso
            w = w + 25*k1/216 + 1408*k3/2565 + 2197*k4/4104 - k5/5                  #usamos RK4 para actualizar la solución
            t = t+h
            ts.append(t)
            sols.append(w)                                                          #guardamos los nuevos valores

        if R == 0:
            q = 4
        else:
            q = 0.84*(tol/R)**(1/4)                                                 #formula para calcular cuanto debemos aumentra o disminuir h

        q = np.clip(q, 0.1, 4)                                                       #obligamos a que q este entre 0.1 y 4
        h = min(hmax, q*h)                                                           #calculamos el nuevo tamaño del paso

    return np.array(ts), np.array(sols)
```
