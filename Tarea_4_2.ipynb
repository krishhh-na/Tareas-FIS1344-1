La matriz de covarianza se define como:

$$
\Sigma = E[(X-\mu)(X-\mu)^T]
$$

Para cualquier vector $$
v =
\begin{pmatrix}
v_1 \\
v_2 \\
\vdots \\
v_n
\end{pmatrix}
$$

Queremos demostrar que $$
v^T \Sigma v \geq 0 $$ ya que una varianza nunca puede ser negativa.
Entonces, reemplazamos la definición de la covarianza en la ecuación:
$$
v^T \Sigma v
= v^T E[(X-\mu)(X-\mu)^T]v
$$

Como $v$ es un vector constante, podemos escribir:

$$
v^T \Sigma v
= E[v^T(X-\mu)(X-\mu)^T v]
$$

Como $v^T(X-\mu)$ es un escalar (un solo número) y $$ (X-\mu)^T v $$ es la transpuesta de esta:
$$ (X-\mu)^T v = v^T(X-\mu)$$
$$v^T(X-\mu)(X-\mu)^T v$$
$$[v^T(X-\mu)]^2$$
$$
v^T \Sigma v
= E[(v^T(X-\mu))^2]
$$

Si tomamos cualquier número real elevado al cuadrado: $$b^2 \geq 0$$ por consiguiente

$$ [(v^T(X-\mu))^2]
$$
Entonces, como no puede ser negativo

$$
E[(v^T(X-\mu))^2] \geq 0
$$

Ahora, interpretando el caso en el cual $$
w^T \Sigma w = 0 $$
para algun vector $w$

Podemos cambiar $v$ por $w$ en el resultado que obtuvimos anteriormente:
$$
w^T \Sigma w
= E[(w^T(X-\mu))^2]
$$
Sabemos que $$E[(w^T(X-\mu))^2] = 0 $$
pero $$(w^T(X-\mu))^2 \geq 0
$$
Para que el promedio de esos cuadrados sea 0, deben ser 0: $$w^T(X-\mu) = 0$$ y en consecuencia $$w^T X - w^T \mu = 0$$
Entonces, $$w^T X = w^T \mu$$
por lo tanto, $$w^T X es constante$$


$$
w^T \Sigma w = Var(w^T X)
$$
