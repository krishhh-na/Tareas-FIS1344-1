de $y" - 2y' + y = te^t -t$





lo pasamos a $y" = 2y' - y + te^t - t$
   

```python
import numpy as np                                     #importamos bibliotecas
import matplotlib.pyplot as plt

def rk4(f, a, b, y0, N):
    h = (b-a)/N                                             #calcula el paso h
    t = a                                                   #condiciones iniciales
    w = np.array(y0, dtype=float)

    ts = np.zeros(N+1)                                      #arreglos para guardar resultados
    sol = np.zeros(N+1)

    ts[0] = t
    sol[0] = w[0]

    def F(t,w):
        return np.append(w[1:], f(t,*w))                    #transformamos la edo en sistema de primer orden

    for i in range(N):                                      #usamos RK4
        k1 = h*F(t,w)
        k2 = h*F(t+h/2, w+k1/2)
        k3 = h*F(t+h/2, w+k2/2)
        k4 = h*F(t+h, w+k3)

        w = w + (k1+2*k2+2*k3+k4)/6                         #actualizamos la solución y tiempo
        t = t+h

        ts[i+1] = t
        sol[i+1] = w[0]

    return ts, sol
```

```python
def f(t,y,dy):
  return 2*dy - y + t*np.exp(t) - t                        #definimos la ecuacion diferencial

a = 0                                                      #condiciones iniciales
b = 1
y0 = [0,0]
N = 100

t, w = rk4(f, a, b, y0, N)                                 #aplicamos rk4

print("aproximación en t = 1:", w[-1])
```

comparamos con la solución exacta:

```python
def exacta(t):                                                             #definimos la solución exacta del problema
  return (t**3/6)*np.exp(t) - t*np.exp(t) + 2*np.exp(t) - t - 2

y_exacta = exacta(t)                                                       #calcular solución exacta
errores = np.abs(w - y_exacta)                                             #error en cada punto

print("resultado numérico:", w[-1])
print("resultado exacto:", y_exacta[-1])
print("error:", np.max(errores))


```

Podemos decir que el método RK4 funciona bien para este problema, ya que si comparamos el resultado numérico con la solución analitica vemos que ambos practicamente coinciden.
