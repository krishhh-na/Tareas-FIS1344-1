```python
datos = np.array([       [5.531683013105e-04,       1.070304113861e+12],
       [5.910227121702e-04,       1.139622191891e+12],
       [6.315688849627e-04,       1.213481232452e+12],
       [6.750065137883e-04,       1.292169597309e+12],
       [7.215301959429e-04,       1.375954296780e+12],
       [7.713235180257e-04,       1.465069697719e+12],
       [8.245525173319e-04,       1.559705507566e+12],
       [8.813589235833e-04,       1.659994890361e+12],
       [9.418537183337e-04,       1.766003776663e+12],
       [1.006111625261e-03,       1.877722471320e+12],
       [1.074167130425e-03,       1.995060685935e+12],
       [1.146012510269e-03,       2.117846703996e+12],
       [1.221598126720e-03,       2.245831037960e+12],
       [1.300834972303e-03,       2.378694410099e+12],
       [1.383599172303e-03,       2.516052095865e+12],
       [1.469737934874e-03,       2.657475892684e+12],
       [1.559076324033e-03,       2.802524407596e+12],
       [1.651424225675e-03,       2.950729179407e+12],
       [1.746582964833e-03,       3.101618970269e+12],
       [1.844351177716e-03,       3.254730780491e+12],
       [1.944529704896e-03,       3.409618695392e+12],
       [2.046925420592e-03,       3.565860320889e+12],
       [2.151354028418e-03,       3.723061333248e+12],
       [2.257641930146e-03,       3.880858018481e+12] ])

import numpy as np
import matplotlib.pyplot as plt
k = datos[:,0]                                 #extraemos las columnas de los datos
p = datos[:,1]
x = np.log(k)                                  #transformamos a escala logartimica
y = np.log(p)

x_uniforme = np.linspace(x[0], x[-1], len(x))  #creamos puntos iguales espaciados en ln(k)
y_uniforme = np.interp(x_uniforme, x, y)

h = x_uniforme[1] - x_uniforme[0]

def deriv3p(f, i, h):                               #formula de tres puntos
    return (f[i+1] - f[i-1])/(2*h)

def deriv3p_extremo(f, i, h, direccion=1):          #formula de tres puntos en los extremos
    return (-3*f[i] + 4*f[i+direccion]
            - f[i+2*direccion])/(2*h*direccion)

def deriv5p(f, i, h):                               #formula de cinco puntos
    return (f[i-2] - 8*f[i-1] + 8*f[i+1]
            - f[i+2])/(12*h)

def deriv5p_extremo(f, i, h, direccion=1):          #formula de cinco puntos en los extremos
    return (-25*f[i] + 48*f[i+direccion]
            - 36*f[i+2*direccion]
            + 16*f[i+3*direccion]
            - 3*f[i+4*direccion])/(12*h*direccion)

n = len(y_uniforme)                                      #creamos los arreglos

D3 = np.zeros(n)
D5 = np.zeros(n)

for i in range(n):

    if i == 0:
        D3[i] = deriv3p_extremo(y_uniforme, i, h)             #derivada de tres puntos
    elif i == n-1:
        D3[i] = deriv3p_extremo(y_uniforme, i, h, -1)
    else:
        D3[i] = deriv3p(y_uniforme, i, h)


    if i < 2:                                                #derivada de cinco puntos
        D5[i] = deriv5p_extremo(y_uniforme, i, h)
    elif i > n-3:
        D5[i] = deriv5p_extremo(y_uniforme, i, h, -1)
    else:
        D5[i] = deriv5p(y_uniforme, i, h)

error = np.abs(D5 - D3)                                      #error, comparamos las derivadas

print("Diferencia absoluta máxima:", np.max(error))
print("Diferencia absoluta media:", np.mean(error))
```

la Diferencia absoluta máxima indica donde se presenta la mayor divergencia entre los dos métodos, la Diferencia absoluta media representa la discrepancia media entre ambas aproximaciones. En nuestro caso, podemos ver que son diferencias muy pequeñas, por lo tanto ambos métodos estan entregando parecidos resultados para la misma derivada.
