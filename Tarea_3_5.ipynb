```python
import random                  #importamos bibliotecas
import numpy as np


def unir(tipo, numero):                                   #creamos las categorias de personas en la aldea
  return {tipo + str(n+1) for n in range(numero)}
aldea = list(unir('V', 50) | unir('M', 30) | unir('N',20)) #definimos que V:siempre dice la verdad, SM:siempre miente y NR:no responde.
len(aldea)


def diez_de_cada_categoria(evento):       #cramos una función que revisa si hay exactamente 10 personas en cada categoria.
  s = [i[0] for i in evento]
  return (s.count('V') == 10 and
          s.count('M') == 10 and
          s.count('N') == 10)

sims = 10                                 #comienza la simulación Monte Carlo
puntos = 200000
probs = np.zeros(sims)

for i in range(sims):
  favorables = 0
  for j in range(puntos):
    muestra = random.sample(aldea, k=30)
    if diez_de_cada_categoria(muestra):
      favorables += 1
      probs[i] = favorables/puntos

probabilidad = probs.mean()                      #pedimos la probabilidad promedio

error = probs.std() / probs.mean()               #pedimos el error relativo


print("Probabilidades obtenidas a partir de la simulación Monte Carlo:")
print(probs)

print("Probabilidad promedio:")
print(probabilidad)

print("Error relativo:")
print(error)

```

Si usamos como criterio mantener el error relativo por debajo de 5%, en nuestro caso 4.93% < 5%, por lo tanto, es un error razonable. Al ser un error relativamente pequeño, nos dice que las distintas simulaciones están entregando resultados buenos o consistentes. Para disminuir este error, debemos aumentar el número de muestras de la simulación
