```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return np.sinh(x)
x_nodos = np.linspace(-5,5,5)
y_nodos = f(x_nodos)
print(x_nodos)
print(y_nodos)

def lagrange(x,puntos):
  resultado = 0
  for i in range(len(puntos)):
    Li = 1

    for j in range(len(puntos)):
      if j != i:
         Li*= (x - puntos[j][0]) / (puntos[i][0] - puntos[j][0])
    resultado += puntos[i][1] * Li
  return resultado
puntos = list(zip(x_nodos, y_nodos))
x = np.linspace(-5,5,500)
P= np.array([lagrange(xi, puntos)for xi in x])

plt.figure(figsize=(10,5))
plt.plot(x, f(x), color='purple', label='f(x) = sinh(x)')
plt.plot(x, P, '--', color='green', label='Polinomio interpolante')
plt.scatter(x_nodos, y_nodos, color='red', label='Nodos')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Interpolación de Lagrange')
plt.grid()
plt.legend()
plt.show()
```

```python
from math import factorial
M = np.cosh(5)
error_estimado = []
for xi in x:
    producto = np.prod([xi - nodo for nodo in x_nodos])
    error_estimado.append(M / factorial(5) * abs(producto))
# Error real
error_real = np.abs(f(x) - P)

# Graficar ambos errores
plt.figure(figsize=(9, 5))

plt.plot(x, error_real, color='red', label='Error real')
plt.plot(x, error_estimado, '--', color='blue' , label='Cota del error')

plt.xlabel('x')
plt.ylabel('Error absoluto')
plt.title('Comparación de errores de interpolación')
plt.grid()
plt.legend()
plt.show()
```
